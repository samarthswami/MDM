```python
# Simulation of Basic Communication System (Tx-Channel-Rx)
import numpy as np
import matplotlib.pyplot as plt

# -------------------------
# Transmitter
# -------------------------

# Time axis
t = np.linspace(0, 1, 1000)

# Original message signal
frequency = 5       # Hz
message = np.sin(2 * np.pi * frequency * t)

# -------------------------
# Communication Channel
# -------------------------

# Generate Gaussian Noise
noise = np.random.normal(0, 0.3, len(t))

# Signal after passing through channel
received_signal = message + noise

# -------------------------
# Receiver
# -------------------------

# In this simple simulation, receiver directly receives noisy signal

recovered_signal = received_signal

# -------------------------
# Plotting
# -------------------------

plt.figure(figsize=(12,8))

plt.subplot(3,1,1)
plt.plot(t, message, color='blue')
plt.title("Original Message Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(3,1,2)
plt.plot(t, noise, color='red')
plt.title("Noise Added by Channel")
plt.xlabel("Time (s)")
plt.ylabel("Noise")
plt.grid(True)

plt.subplot(3,1,3)
plt.plot(t, recovered_signal, color='green')
plt.title("Received Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate binary data
N = 20
tx = np.random.randint(0, 2, N)

# BPSK Modulation
signal = 2*tx - 1

# Add Gaussian Noise
noise = np.random.normal(0, 0.4, N)
received = signal + noise

# Detection
rx = (received > 0).astype(int)

print("Transmitted:", tx)
print("Received:", rx)

plt.figure(figsize=(8,6))

plt.subplot(3,1,1)
plt.stem(tx)
plt.title("Transmitted Bits")

plt.subplot(3,1,2)
plt.plot(received,'o-')
plt.grid(True)
plt.title("Signal after Channel")

plt.subplot(3,1,3)
plt.stem(rx)
plt.title("Recovered Bits")

plt.tight_layout()
plt.show()
```
