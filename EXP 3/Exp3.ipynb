```python
import matplotlib.pyplot as plt
from matplotlib.patches import RegularPolygon
import numpy as np

# ------------------------------------
# Hexagon parameters
# ------------------------------------
R = 1.0
dx = 1.5 * R
dy = np.sqrt(3) * R

# ------------------------------------
# Colors for the 7 cells
# ------------------------------------
cell_colors = {
    'A': 'red',
    'B': 'blue',
    'C': 'green',
    'D': 'orange',
    'E': 'purple',
    'F': 'cyan',
    'G': 'yellow'
}

# ------------------------------------
# Seven cells in one frequency cluster
# ------------------------------------
cluster = [
    (0,       dy,    'B'),   # Top
    (-dx,     dy/2,  'G'),   # Upper-left
    ( dx,     dy/2,  'C'),   # Upper-right
    (0,       0,     'A'),   # Centre
    (-dx,    -dy/2,  'F'),   # Lower-left
    ( dx,    -dy/2,  'D'),   # Lower-right
    (0,      -dy,    'E')    # Bottom
]

# ------------------------------------
# Position of the three 7-cell clusters
# ------------------------------------
cluster_offsets = [
    (0, 0),                  # Upper cluster
    (-3*R, -2*dy),          # Lower-left cluster
    (1.5*R, -2.5*dy)        # Lower-right cluster
]

# ------------------------------------
# Create all cells
# ------------------------------------
cells = []

for ox, oy in cluster_offsets:
    for x, y, label in cluster:
        cells.append((x + ox, y + oy, label))

# ------------------------------------
# Draw hexagons
# ------------------------------------
fig, ax = plt.subplots(figsize=(8, 5.8))

for x, y, label in cells:

    hexagon = RegularPolygon(
        (x, y),
        numVertices=6,
        radius=R,
        orientation=np.radians(30),
        facecolor=cell_colors[label],
        edgecolor='black',
        linewidth=2.0,
        alpha=0.7
    )

    ax.add_patch(hexagon)

    # Cell label
    ax.text(
        x, y, label,
        ha='center',
        va='center',
        fontsize=11,
        color='black',
        fontweight='bold'
    )

# ------------------------------------
# Figure settings
# ------------------------------------
ax.set_aspect('equal')

ax.set_xlim(-6, 6)
ax.set_ylim(-7.5, 3)
ax.axis('off')

# ------------------------------------
# Title
# ------------------------------------
ax.text(
    -5.1, 2.65,
    'Frequency Reuse',
    ha='left',
    va='top',
    fontsize=11
)

plt.tight_layout(pad=0.2)
plt.show()
```
